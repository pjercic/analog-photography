# ⚗️ Chemistry: Archival Print Processing

### Welcome, Darkroom Archivists!
[cite_start]It is comforting for photographers to know that their images will survive the ravages of time to become an important legacy for the next generation[cite: 2]. [cite_start]While modern inkjet prints often claim archival status based on accelerated testing, carbon-based monochrome prints can visibly fade within six months[cite: 433]. 

[cite_start]To achieve true permanence, fiber-base (FB) prints, combined with a carefully controlled full archival process, have the best chance of survival[cite: 7]. 

[cite_start]In short, archival processing requires three critical steps[cite: 11]:
1. **Fixing:** Removing all unexposed silver.
2. **Toning:** Protecting the remaining image silver.
3. **Washing:** Thoroughly removing harmful chemicals from the emulsion and paper fibers.

## Step 1: The Two-Bath Fixing Method
[cite_start]During development, exposed silver halides are reduced to metallic silver, but the unexposed silver halides remain light-sensitive and must be removed[cite: 14, 15]. [cite_start]Fixers use thiosulfate to dissolve these residual silver halides without damaging the metallic image[cite: 29].

[cite_start]However, as the fixer dissolves silver, it becomes contaminated with soluble silver thiosulfate complexes[cite: 30]. If these complexes build up too high, they will stain and destroy the print over time. 

To prevent this, archival printers use the **Two-Bath Method**:
* **Bath 1:** Does the heavy lifting. [cite_start]It is used until its silver thiosulfate content reaches 0.5 to 1.0 g/l[cite: 97]. 
* **Bath 2:** Acts as an insurance policy. [cite_start]A fresher second bath ensures any remaining silver halides and insoluble complexes are completely rendered soluble[cite: 32]. 

Let's simulate how the contamination builds up in these two baths over time!

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_two_bath_fixing():
    """
    Simulates the silver thiosulfate contamination in a two-bath fixing system.
    """
    prints_processed = np.arange(0, 50)
    
    # Bath 1 takes the brunt of the unexposed silver
    # Reaches archival limit (1.0 g/l) at roughly 20 prints per liter [cite: 98]
    bath_1_contamination = prints_processed * (1.0 / 20)
    
    # Bath 2 only processes the tiny amount of silver Bath 1 missed
    bath_2_contamination = prints_processed * (0.05 / 20)
    
    plt.figure(figsize=(10, 6))
    plt.plot(prints_processed, bath_1_contamination, 'r-', lw=3, label="Bath 1 (Primary Fixer)")
    plt.plot(prints_processed, bath_2_contamination, 'b-', lw=3, label="Bath 2 (Secondary Fixer)")
    
    # Archival safety limit
    plt.axhline(1.0, color='k', linestyle='--', label="Archival Limit (1.0 g/l) [cite: 97]")
    
    plt.title("Two-Bath Fixing: Silver Contamination over Time", fontsize=16)
    plt.xlabel("Number of 8x10 Prints Processed")
    plt.ylabel("Silver Thiosulfate Concentration (g/l)")
    plt.grid(True, alpha=0.4)
    plt.legend()
    plt.show()

# Run the fixer simulation!
simulate_two_bath_fixing()
```

## Step 2: Toning for Protection
[cite_start]Even a perfectly fixed print is vulnerable to environmental oxidants[cite: 336]. [cite_start]Toning converts the image-forming metallic silver into more inert silver compounds, guarding against premature deterioration[cite: 109]. 


* [cite_start]**Sulfide Toning:** Effectively converts metallic silver to the far more stable silver sulfide[cite: 136]. 
* [cite_start]**Selenium Toning:** Converts silver to inert silver selenide, gently darkening shadows and midtones[cite: 179, 180]. 
* [cite_start]**Combination Toning:** A blend of selenium and polysulfide protects all print tones, achieving maximum archival stability[cite: 200].

## Step 3: Washing (The Physics of Diffusion)
A fixed print is saturated with thiosulfate. [cite_start]The process of print washing is a combination of displacement and diffusion[cite: 221]. 

[cite_start]As long as there is a difference in thiosulfate concentration between the print and the wash water, thiosulfate will diffuse from the print into the water[cite: 227]. [cite_start]However, diffusion is an exponential process that decreases geometrically with time, meaning it slows down rapidly as it approaches equilibrium[cite: 234, 235]. 

[cite_start]Because paper fibers have a tendency to adsorb residual thiosulfate, we use a **Washing Aid** (hypo-clearing agent)[cite: 239, 241]. [cite_start]It desorbs the thiosulfate, reducing the total processing time by about 50%[cite: 247].

Let's use Python to graph the exponential decay of washing with and without a washing aid!

```python
def simulate_washing_diffusion():
    """
    Simulates the exponential diffusion of thiosulfate out of a print,
    comparing a plain water wash to a wash using a Washing Aid.
    """
    time_minutes = np.linspace(0, 60, 100)
    
    # Exponential decay formulas for diffusion
    # Plain wash is sluggish because paper fibers hold onto thiosulfate [cite: 239]
    plain_wash_residual = 100 * np.exp(-0.05 * time_minutes)
    
    # Washing aid dramatically speeds up the desorption [cite: 247]
    aid_wash_residual = 100 * np.exp(-0.15 * time_minutes)
    
    plt.figure(figsize=(10, 6))
    plt.plot(time_minutes, plain_wash_residual, 'r-', lw=2, label="Plain Water Wash")
    plt.plot(time_minutes, aid_wash_residual, 'g-', lw=3, label="Wash with Washing Aid")
    
    # Target archival limit (near zero)
    plt.axhline(5, color='k', linestyle='--', label="Archival Target Level")
    
    plt.title("Print Washing: The Geometry of Diffusion", fontsize=16)
    plt.xlabel("Washing Time (Minutes)")
    plt.ylabel("Residual Thiosulfate (%)")
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend()
    plt.show()

# Run the washing diffusion simulator!
simulate_washing_diffusion()
```
