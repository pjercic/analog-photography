# Framing and Displaying Prints: A Pedagogical Guide

Once your print is beautifully mounted and matted, the final step in its journey is framing and exhibition. [cite_start]A frame serves three essential purposes: it isolates the print from its surroundings, enhances it aesthetically, and protects it against dirt, dust, and rough handling[cite: 515]. 

However, a frame should always act as a supporting player. [cite_start]It must complement the print without competing with it for attention—you want people to exhibit the print, not the frame[cite: 515].

Let's begin by looking at the materials used to build frames. [cite_start]While wood frames are popular and traditional, they contain aromatic oils that vaporize over time and can attack your photographs[cite: 517]. [cite_start]Paints and varnishes emit similar harmful fumes[cite: 517]. [cite_start]For absolute archival safety, chemically inert plastics and unpainted metals are the best choices[cite: 517]. [cite_start]Extruded aluminum is the professional material of choice because it is light, easy to cut, and resists corrosion without the need for paint [cite: 517-518].

Let's model a simple quality-control check for frame materials in Python.

```python
class FrameMaterial:
    """
    A pedagogical class to evaluate the archival safety of framing materials.
    """
    def __init__(self, material_type, has_paint_or_varnish=False):
        self.material_type = material_type.lower()
        self.has_paint_or_varnish = has_paint_or_varnish
        
    def evaluate_safety(self):
        print(f"\n--- Evaluating Frame Material: {self.material_type.capitalize()} ---")
        
        if self.has_paint_or_varnish:
            print("WARNING: Paints and varnishes emit harmful fumes that can degrade prints over time.")
            
        if self.material_type == "wood":
            print("STATUS: Not Archival. Wood contains aromatic oils that vaporize and attack photographs.")
        elif self.material_type in ["aluminum", "metal"]:
            print("STATUS: Highly Recommended. Aluminum is light, corrosion-resistant, and chemically inert.")
        elif self.material_type == "plastic":
            print("STATUS: Archival, provided the plastic is chemically inert. However, often considered aesthetically inferior.")
        else:
            print("STATUS: Unknown material. Stick to unpainted metal for maximum archival safety.")

# Let's test a traditional painted wood frame versus an unpainted aluminum frame
wood_frame = FrameMaterial("wood", has_paint_or_varnish=True)
wood_frame.evaluate_safety()

aluminum_frame = FrameMaterial("aluminum", has_paint_or_varnish=False)
aluminum_frame.evaluate_safety()
```

## Step 2: The Science of Glazing (Glass vs. Plastic)

[cite_start]A print is put behind glass for protection, but this protection comes at an aesthetic cost[cite: 518]. [cite_start]Any glazing material will slightly change the image color and reduce image contrast[cite: 521]. 

If a perfect glazing material existed, it would transmit 100% of light. [cite_start]In reality, standard float glass reflects about 8% of the light and absorbs another 2%, leaving only about 90% of the light to actually reach and transmit from your print[cite: 519]. [cite_start]Because highlights naturally reflect a lot of light and shadows reflect very little, losing 10% of the total transmitted light reduces the brightness of your highlights much more severely than your shadows[cite: 521]. [cite_start]This results in an overall drop in image contrast[cite: 521].

You have a few choices for glazing:
* [cite_start]**Standard Float Glass:** Inexpensive, scratch-resistant, perfectly flat, and easy to clean, transmitting about 90% of light[cite: 524].
* [cite_start]**Acrylic (Plexiglas):** Lighter and shatter-resistant, but attracts dust, is hard to clean, and is rarely perfectly flat [cite: 523-524].
* [cite_start]**Optical Glass:** Coated to transmit 99% of light with almost no reflections, but prohibitively expensive for everyday use[cite: 525].

Let's write a program to calculate how standard glass reduces the contrast of a print by simulating this loss of light transmission.

```python
class GlazingCalculator:
    """
    Calculates the impact of glazing on print contrast.
    """
    def __init__(self, glazing_type, transmittance):
        self.glazing_type = glazing_type
        self.transmittance = transmittance # Represented as a decimal (e.g., 0.90 for 90%)
        
    def calculate_contrast_loss(self, highlight_reflectance, shadow_reflectance):
        print(f"\n--- Glazing Analysis: {self.glazing_type} ---")
        print(f"Glazing Transmittance: {self.transmittance * 100}%")
        
        # Calculate original contrast ratio
        original_contrast = highlight_reflectance / shadow_reflectance
        
        # Calculate perceived reflectance through the glazing
        perceived_highlight = highlight_reflectance * self.transmittance
        perceived_shadow = shadow_reflectance * self.transmittance
        
        # The absolute brightness of the highlight drops significantly
        highlight_drop = highlight_reflectance - perceived_highlight
        print(f"Highlight brightness drops by {highlight_drop:.1f} units.")
        
        # In actual practice, ambient light reflections on the front of the glass 
        # heavily wash out the shadows, further lowering the effective contrast ratio.
        print("Note: Because absolute highlight brightness is reduced, the overall perceived contrast is flattened.")

# Let's simulate a standard float glass (90% transmittance) 
# on a print with bright highlights (81% reflectance) and deep shadows (1% reflectance).
standard_glass = GlazingCalculator("Standard Float Glass", 0.90)
standard_glass.calculate_contrast_loss(81.0, 1.0)
```

## Step 3: Assembly and The Golden Rule of Framing

When assembling your frame, you need a profile deep enough to hold all your layers. [cite_start]A popular professional choice is the Nielsen #15 aluminum profile, which has a 15 mm pocket[cite: 518]. [cite_start]This easily accommodates your glass (approx. 2 mm), your overmat, your mount-board, and a sturdy 1/4-inch foam backboard[cite: 518].

[cite_start]**The Golden Rule:** Regardless of the glazing you select, your photographic print must **never** be placed directly against the glass surface[cite: 525]. 

[cite_start]If the print touches the glass, normal humidity will cause the gelatin emulsion of the photograph to swell and stick permanently to the glazing[cite: 525]. [cite_start]If this happens, separating the print from the glass is nearly impossible, and the print is usually ruined[cite: 525]. [cite_start]The **overmat** acts as a crucial spacer, providing a physical air gap between the print and the glass[cite: 525].

Let's model the assembly process to ensure our frame pocket is deep enough and that our golden rule is strictly followed.

```python
class FrameAssembly:
    def __init__(self, profile_depth_mm):
        self.profile_depth_mm = profile_depth_mm
        self.layers = []
        self.has_overmat = False
        self.has_glass = False
        
    def add_layer(self, name, thickness_mm):
        self.layers.append({"name": name, "thickness": thickness_mm})
        if "overmat" in name.lower():
            self.has_overmat = True
        if "glass" in name.lower():
            self.has_glass = True
            
    def check_assembly(self):
        print(f"\n--- Checking Frame Assembly ---")
        total_thickness = sum(layer["thickness"] for layer in self.layers)
        
        print(f"Total layer thickness: {total_thickness:.1f} mm")
        print(f"Available frame depth: {self.profile_depth_mm:.1f} mm")
        
        if total_thickness > self.profile_depth_mm:
            print("ERROR: Your layers are too thick for this frame profile!")
            return False
            
        if self.has_glass and not self.has_overmat:
            print("CRITICAL ERROR: Golden Rule Violation! You have glass but no overmat.")
            print("The gelatin emulsion will stick to the glass and ruin the print.")
            return False
            
        print("SUCCESS: Assembly fits perfectly and the print is protected by the overmat air gap.")
        return True

# Let's assemble a frame using the standard Nielsen #15 profile (15mm depth)
my_frame = FrameAssembly(profile_depth_mm=15.0)

# Add our standard layers
my_frame.add_layer("Standard Float Glass", 2.0)
my_frame.add_layer("Cotton Rag Overmat", 1.5)
my_frame.add_layer("Silver-Gelatin Print", 0.3)
my_frame.add_layer("Mount-Board", 1.5)
my_frame.add_layer("Foam Backboard", 6.35) # 1/4 inch is approx 6.35mm

my_frame.check_assembly()
```

## Step 4: Exhibition and Panel Design

Once your prints are framed, you are ready to exhibit! But hanging prints together requires careful curation. 

[cite_start]A "panel" is a coherent display of a number of images[cite: 539]. [cite_start]To design a successful panel, each print must fit a common theme[cite: 539]. [cite_start]Crucially, the presentation must be uniform: the prints must be printed, toned, mounted, and framed in a matching style[cite: 539]. [cite_start]Mixing different toning styles, mount-boards, or frames within one panel looks amateurish[cite: 539]. [cite_start]A well-designed panel is a cohesive piece of art in itself[cite: 539].
